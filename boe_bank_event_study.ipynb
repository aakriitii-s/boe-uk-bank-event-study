```python
def calculate_event_study(event_date, stock_returns, stock_ticker, market_ticker="^FTSE"):
    
    event_position = stock_returns.index.get_loc(event_date)
    
    # Need 250+ trading days before the event
    estimation_start = event_position - 270
    estimation_end = event_position - 6
    
    # Need a complete -5 to +5 event window
    event_start = event_position - 5
    event_end = event_position + 5
    
    # Check that the full event window exists
    if event_start < 0 or event_end >= len(stock_returns):
        return None
    
    estimation = stock_returns.iloc[
        estimation_start:estimation_end + 1
    ]
    
    X = sm.add_constant(estimation[market_ticker])
    y = estimation[stock_ticker]
    
    model = sm.OLS(y, X).fit()
    
    alpha = model.params["const"]
    beta = model.params[market_ticker]
    
    event_data = stock_returns.iloc[
        event_start:event_end + 1
    ].copy()
    
    market_returns = event_data[market_ticker]
    actual_returns = event_data[stock_ticker]
    
    expected_returns = alpha + beta * market_returns
    
    abnormal_returns = actual_returns - expected_returns
    
    event_data["Expected_Return"] = expected_returns
    event_data["Abnormal_Return"] = abnormal_returns
    event_data["Event_Day"] = range(-5, 6)
    
    car_minus1_plus1 = abnormal_returns.iloc[4:7].sum()
    car_minus5_plus5 = abnormal_returns.sum()
    
    return event_data, alpha, beta, car_minus1_plus1, car_minus5_plus5
```

```python
confidence_results = []

for bank in banks:
    
    data = results_df[
        (results_df["Decision"] == "Hike") &
        (results_df["Bank"] == bank)
    ]["CAR_-1_+1"]
    
    mean = data.mean()
    sem = stats.sem(data)
    
    confidence_interval = stats.t.interval(
        0.95,
        df=len(data)-1,
        loc=mean,
        scale=sem
    )
    
    confidence_results.append({
        "Bank": bank,
        "Mean_CAR_%": mean * 100,
        "CI_Lower_%": confidence_interval[0] * 100,
        "CI_Upper_%": confidence_interval[1] * 100
    })

confidence_results = pd.DataFrame(confidence_results)

confidence_results
```

```python
from scipy import stats

aar_tests = []

for day in sorted(event_time_full["Event_Day"].unique()):
    
    values = event_time_full.loc[
        event_time_full["Event_Day"] == day,
        "Abnormal_Return"
    ].dropna()
    
    t_stat, p_value = stats.ttest_1samp(values, 0)
    
    aar_tests.append({
        "Event_Day": day,
        "AAR": values.mean(),
        "AAR_%": values.mean() * 100,
        "t_stat": t_stat,
        "p_value": p_value,
        "N": len(values)
    })

aar_tests = pd.DataFrame(aar_tests)

aar_tests
```

```python
decision_tests = []

for decision in ["Hike", "Cut", "Hold"]:
    
    values = event_time_full.loc[
        event_time_full["Decision"] == decision
    ].groupby("Event_Date")["Abnormal_Return"].sum()
    
    t_stat, p_value = stats.ttest_1samp(values, 0)
    
    decision_tests.append({
        "Decision": decision,
        "Mean_CAR": values.mean(),
        "Mean_CAR_%": values.mean() * 100,
        "t_stat": t_stat,
        "p_value": p_value,
        "N": len(values)
    })

decision_tests = pd.DataFrame(decision_tests)

decision_tests
```
